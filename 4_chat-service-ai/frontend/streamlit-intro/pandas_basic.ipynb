# 판다스란?
- 차트 그리기용
- 다양한 타입의 데이터를 표로 다루는 애
- 표 또는 차트 그릴 수 있도록 도와줌
- 벡터라이제이션 vectorization (리스트로 데이터를 가진 일차원 프레임 구조)


```python
import pandas as pd
```

```python
df_simple = pd.DataFrame({
    "과일": ["사과", "바나나", "딸기", "포도"],
    "가격": [1500, 800, 2000, 3500],
    "수량": [10, 25, 15, 8],
})

df_simple
```

```python
df_simple.describe() #기술통계
```

```python
#원하는 컬럼만 뽑아낼 수 있다

df_simple["가격"] #시리즈 꺼내는 것
```

```python
# 타입 변경 가능
df_simple["가격"].astype(str)
```

```python
df_simple.info()  # 위 타입 변경은 복사본에 한것이라 원본은 타입변경 되지않음
```

```python
df_simple["가격"] = df_simple["가격"].astype(str) # 이렇게 원본에 넣어줘야 타입 변경 완료

df_simple.info()
```

```python
# 리스트로 시리즈 생성
data = [1, 2, 3, 4, 5]
s1 = pd.Series(data) # s1 이 시리즈
print(s1)

type(s1)
```

```python
# 인덱스 지정하여 시리즈 생성
data = [1, 2, 3, 4, 5]
s_name = ['a', 'b', 'c', 'd', 'e']
s2 = pd.Series(data, index=s_name)
print(s2)
```

```python
# 딕셔너리로 시리즈 생성 - 키가 인덱스가 됨
dict_data = {'a':1, 'b':2, 'c':3, 'd':4}
s3 = pd.Series(dict_data)
print(s3)
```

```python
# location - 판다시의 시리즈인걸 한눈에 알아볼 수 있어서 이롷게 쓰는걸 선호한다
# 인덱스명을 사용한다
# 라벨 인데스
s3.loc['a'] 
```

```python
#index의 loc, 자동생성 index 숫자를 사용한다
#위치 인덱싱 - 항상 0으로 시작함
s3.iloc['a'] 
```

```python
population = pd.Series([9776, 2729, 1566, 3158], 
                    index=['서울', '부산', '인천', '대구'])

print(population['서울'])  # 9776 출력
print(population[['서울', '부산']])  # 서울과 부산의 데이터만 선택
```

```python
print(population['인천'])
print(population.loc['인천'])
print(population.iloc[2])
```

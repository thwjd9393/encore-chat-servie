```python
# TODO 1. "day14:greeting" 에 "안녕하세요" 를 저장한다.  힌트: r.set(키, 값)
...
r.set("day14:greeting","안녕하세요")
print("저장한 값:", r.get("day14:greeting"))
```

## 2. Hash - 항목이 여러 개인 정보

```python
# json 형태는 hash로 관리한다

# 사용자 한 명의 정보를 한 키에 담는다.
# 해시는 필드 값을 넣을 수 있다!!!

# TODO 3. Hash 로 name 을 저장한다.  힌트: r.hset(키, 필드, 값)
r.hset("day14:user", "name", "김철수")
r.hset("day14:user", "email", "kim@example.com")
r.hset("day14:user", "point", 100)

print("이름만 꺼내기:", r.hget("day14:user", "name"))
print("전체 꺼내기:", r.hgetall("day14:user"))

```

## 7. 정리

keys() 대신 scan_iter()를 쓰는 편이 좋다
결과는 같지만 keys("*")는 키가 많을 때 Redis 전체를 잠시 멈춤 *************

```python
keys = r.keys("day14:*")
for key in keys:
    r.delete(key)

print(len(keys), "개 삭제")
print("남은 키 개수:", r.dbsize())
```

```python
keys = list(r.scan_iter("day14:*"))
for key in keys:
    r.delete(key)

print(len(keys), "개 삭제")
print("남은 키 개수:", r.dbsize())
```
